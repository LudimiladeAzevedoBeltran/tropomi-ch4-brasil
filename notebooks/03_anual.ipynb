# 3) Série anual de CH4 no Brasil — TROPOMI

A escala **anual** é derivada da série mensal (média dos meses válidos de cada
ano). É mais robusta do que depender de `P1Y` na API — veja a página *Anual* no
site para a discussão.


```python
import sys
from pathlib import Path

# Deixa o pacote tropomi_ch4 visível quando o notebook roda de notebooks/
sys.path.insert(0, str(Path.cwd().parent))
# Se tiver instalado com `pip install -e .`, as duas linhas acima são dispensáveis.

import tropomi_ch4 as tch4
from tropomi_ch4 import plot

print("tropomi_ch4", tch4.__version__)
```

## Rodar a série anual (via mensal)

```python
start = "2019-01-01"
end = None

df = tch4.annual(start=start, end=end)
df
```

## Salvar o CSV e a figura

```python
out = Path.cwd().parent / "data"
out.mkdir(exist_ok=True)
csv = out / "serie_anual_ch4_brasil.csv"
df.to_csv(csv, index=False)
print("salvo:", csv)

plot.plot_series(
    df,
    title="CH4 médio anual no Brasil — TROPOMI",
    outfile=out / "serie_anual_ch4_brasil.png",
)
```
