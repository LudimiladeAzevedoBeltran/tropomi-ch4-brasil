# 1) Série diária de CH4 no Brasil — TROPOMI

Média **diária** (`P1D`) de CH4 sobre o Brasil, via Statistical API do
Copernicus Data Space (sem baixar imagens).

- Antes de rodar, preencha o `.env` (veja o site: *Conta Copernicus* e *Links da API*).
- A série enfrenta dias sem observação; a cobertura sobe ao longo dos anos.


```python
import sys
from pathlib import Path

# Deixa o pacote tropomi_ch4 visível quando o notebook roda de notebooks/
sys.path.insert(0, str(Path.cwd().parent))
# Se tiver instalado com `pip install -e .`, as duas linhas acima são dispensáveis.

import tropomi_ch4 as tch4
from tropomi_ch4 import plot

print("tropomi_ch4", tch4.__version__)
```

## Rodar a série diária

A função `daily` faz uma requisição por ano (mais robusta a falhas).

```python
# Ajuste o período se quiser
start = "2019-01-01"
end = None            # None = até hoje

df = tch4.daily(start=start, end=end)
df.head()
```

## Salvar o CSV

```python
out = Path.cwd().parent / "data"
out.mkdir(exist_ok=True)
csv = out / "serie_diaria_ch4_brasil.csv"
df.to_csv(csv, index=False)
print("salvo:", csv, "| dias:", len(df), "| com dado:", int(df["ch4_ppb"].notna().sum()))
```

## Figura

```python
plot.plot_series(
    df,
    title="CH4 médio diário no Brasil — TROPOMI",
    outfile=out / "serie_diaria_ch4_brasil.png",
)
```
