# 4) Mapas com o L2 cru (~1 km)

Para **mapas** mensais em ~1 km é preciso o produto **L2 cru**.

> ⚠️ **IMPORTANTE — como o L2 funciona:** o processo baixa o **granule completo**
> (um arquivo de ~70 MB, cobrindo uma faixa do mundo), usa **apenas os pixels da
> região do Brasil** e **APAGA o arquivo completo** em seguida.
>
> Ou seja: **o arquivo cheio é temporário**. Fica salvo só o **GeoTIFF mensal já
> recortado**. Por isso o **disco usado é pequeno** — existe 1 granule por vez.
>
> O custo real é **banda/tempo**: ~6 GB e ~20–25 min **por mês**. Comece com um
> mês de teste e só depois amplie.


```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import tropomi_ch4 as tch4
from tropomi_ch4 import plot, l2

print("tropomi_ch4", tch4.__version__)
```

## 1) Listar os granules do mês

Só lista (não baixa nada). Em 2022-01 há ~93 granules.

```python
prods = l2.list_granules(2022, 1)
print("granules:", len(prods))
prods[:2]
```

## 2) Testar um mês, com poucos granules

`max_granules=5` limita o download. Remova (ou use `None`) para o mês todo.

Cada granule baixado é **recortado para o Brasil** e depois **excluído**; ao fim,
sobra só o GeoTIFF `CH4_2022_01.tif`.

```python
out = Path.cwd().parent / "data"
out.mkdir(exist_ok=True)

media = l2.build_monthly_geotiff(
    year=2022, month=1,
    out_tif=out / "CH4_2022_01.tif",
    res=0.01,          # ~1 km
    max_granules=5,    # teste; None = mês todo
)
print("CH4 médio (ppb):", round(media, 2))
```
